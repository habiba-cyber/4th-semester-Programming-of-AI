```python
# Python Programming Fundamentals

# Function to display a welcome message
def welcome_user(user_name):
    return "Welcome, " + user_name

print(welcome_user("John"))


# Decision Making using if-elif-else

value = 8

if value > 10:
    print("Value is greater than 10")
elif value == 10:
    print("Value is exactly 10")
else:
    print("Value is smaller than 10")


# Loop Demonstrations

# Using for loop
print("Demonstration of For Loop:")
for num in range(1, 6):
    print("Current number:", num)

# Using while loop
print("Demonstration of While Loop:")
counter = 1
while counter <= 5:
    print("Step:", counter)
    counter += 1


# Python Data Structures

# List example
colors = ["red", "blue", "green"]
colors.append("yellow")
print("Color List:", colors)


# Tuple example
point = (5, 9)
print("First coordinate:", point[0])


# Set example
numbers_set = {2, 4, 6, 6, 8}
print("Unique Numbers:", numbers_set)


# Dictionary example
person = {
    "username": "John123",
    "score": 95
}

print("Username:", person["username"])
print("Score:", person["score"])


# Small Program Example

# Function to check positive or negative number
def check_sign(n):
    if n >= 0:
        return "This number is positive"
    else:
        return "This number is negative"

test_value = -4
print("Test Value:", test_value)
print(check_sign(test_value))


# Loop through list elements
print("Available Colors:")
for color in colors:
    print(color)


# Loop through dictionary
print("Person Details:")
for key, val in person.items():
    print(key, "->", val)
```
