```python
JUG_A_CAP = 4
JUG_B_CAP = 3
TARGET = 2

def dfs_search(current_state, path, visited):
    a, b = current_state
    if a == TARGET or b == TARGET:
        path.append(current_state)
        return path
    
    visited.add(current_state)
    path.append(current_state)
    
    next_moves = []
    
    next_moves.append((JUG_A_CAP, b))
    next_moves.append((a, JUG_B_CAP))
    next_moves.append((0, b))
    next_moves.append((a, 0))
    
    pour = min(a, JUG_B_CAP - b)
    next_moves.append((a - pour, b + pour))
    
    pour = min(b, JUG_A_CAP - a)
    next_moves.append((a + pour, b - pour))
    
    for next_state in next_moves:
        if next_state not in visited:
            result = dfs_search(next_state, path.copy(), visited.copy())
            if result is not None:
                return result
    
    return None

start_state = (0, 0)
visited_states = set()
solution = dfs_search(start_state, [], visited_states)

for i in range(len(solution)-1):
    a1, b1 = solution[i]
    a2, b2 = solution[i+1]
    
    if a2 > a1:
        print(f"Step {i+1}: Fill Jug A -> {solution[i+1]}")
    elif b2 > b1:
        print(f"Step {i+1}: Fill Jug B -> {solution[i+1]}")
    elif a2 < a1 and b2 == b1:
        print(f"Step {i+1}: Empty Jug A -> {solution[i+1]}")
    elif b2 < b1 and a2 == a1:
        print(f"Step {i+1}: Empty Jug B -> {solution[i+1]}")
    elif a2 > a1 and b2 < b1:
        print(f"Step {i+1}: Pour Jug B -> Jug A -> {solution[i+1]}")
    elif a2 < a1 and b2 > b1:
        print(f"Step {i+1}: Pour Jug A -> Jug B -> {solution[i+1]}")

print("Goal reached:", solution[-1])
```
