```python

import cv2
from matplotlib import pyplot as plt
import numpy as np
from tkinter import Tk
from tkinter.filedialog import askopenfilename

# Open file dialog to select image
Tk().withdraw()  # Hide the root window
image_path = askopenfilename(title="Select an Image")  # Pop-up to select your image

if not image_path:
    raise FileNotFoundError("No image selected. Please run the cell again and choose an image!")

# Step 2: Read image
image = cv2.imread(image_path)

if image is None:
    raise ValueError("Failed to load image. Make sure it's a valid image file.")

# Step 3: Convert BGR to RGB for Matplotlib
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# --- Step 4: Display original image ---
plt.figure(figsize=(6,6))
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis('off')
plt.show()

# Step 5: Resize image
resized = cv2.resize(image, (300, 300))
resized_rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)
plt.imshow(resized_rgb)
plt.title("Resized Image")
plt.axis('off')
plt.show()

# Step 6: Convert to Grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis('off')
plt.show()

# Step 7: Draw Shapes
shapes_img = image.copy()
cv2.line(shapes_img, (50,50), (200,50), (255,0,0), 3)
cv2.rectangle(shapes_img, (50,70), (200,150), (0,255,0), 2)
cv2.circle(shapes_img, (150,300), 50, (0,0,255), -1)
cv2.putText(shapes_img, 'OpenCV Demo', (50,400), cv2.FONT_HERSHEY_SIMPLEX, 1, (255,255,0), 2)

shapes_rgb = cv2.cvtColor(shapes_img, cv2.COLOR_BGR2RGB)
plt.imshow(shapes_rgb)
plt.title("Shapes Drawn")
plt.axis('off')
plt.show()

# Step 8: Edge Detection
edges = cv2.Canny(gray, 100, 200)
plt.imshow(edges, cmap='gray')
plt.title("Edge Detection")
plt.axis('off')
plt.show()

# Step 9: Contour Detection
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
contours, _ = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
contour_img = image.copy()
cv2.drawContours(contour_img, contours, -1, (0,255,255), 2)
contour_rgb = cv2.cvtColor(contour_img, cv2.COLOR_BGR2RGB)
plt.imshow(contour_rgb)
plt.title("Contours")
plt.axis('off')
plt.show()

# Step 10: Face Detection
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5)
face_img = image.copy()
for (x, y, w, h) in faces:
    cv2.rectangle(face_img, (x,y), (x+w,y+h), (255,0,255), 2)
face_rgb = cv2.cvtColor(face_img, cv2.COLOR_BGR2RGB)
plt.imshow(face_rgb)
plt.title("Faces Detected")
plt.axis('off')
plt.show()
```
