```python
class NQueens:
    def __init__(self, N):
        self.N = N
        self.board = [-1] * N  # Dynamic board: index=row, value=column

    def is_safe(self, row, col):
        for prev_row in range(row):
            # Check column and diagonals
            if self.board[prev_row] == col or \
               abs(self.board[prev_row] - col) == abs(prev_row - row):
                return False
        return True

    def solve_nqueens(self, row=0):
        if row == self.N:
            self.print_board()
            return True

        solved = False
        for col in range(self.N):
            if self.is_safe(row, col):
                self.board[row] = col
                solved = self.solve_nqueens(row + 1) or solved
                self.board[row] = -1  # Backtrack
        return solved

    def print_board(self):
        for row in range(self.N):
            line = ['Q' if self.board[row] == col else '.' for col in range(self.N)]
            print(" ".join(line))
        print("\n" + "-" * (self.N * 2))

# Example usage
n = 4
solver = NQueens(n)
solver.solve_nqueens()
```
